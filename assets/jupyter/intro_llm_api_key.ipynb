In order to make a call to run LLM inference, you will need an API key. You can get an API key from a number of different inference providers, some free and some paid.

For free options, there is access to an API key for researchers in the US available after obtaining a namespace (https://nrp.ai/llmtoken/). Google also offers inference through an API key with limited rates. You can generate a free API key through Google AI Studio without entering payment information by going to (https://aistudio.google.com/app/api-keys) and clicking the "Create API key" button.

Set the provider's base URL and then paste your key when prompted. The key is hidden and is not saved in the notebook. (For Gemini, use `https://generativelanguage.googleapis.com/v1beta/openai/` as the base URL.)

```python
from getpass import getpass
from openai import OpenAI

base_url = "PASTE_BASE_URL_HERE"
api_key = getpass("API key: ")

client = OpenAI(base_url=base_url, api_key=api_key)
```

List the available models, then copy one name into `MODEL` below.

**Remark**: If you are using the free Gemini API, some models in the returned list will not be accessible to you; you can still try to run the inference with these models but you will receive an error. Try looking up which models are provided for free, or which models are provided through your provider and at what price.

```python
for model in client.models.list():
    print(model.id)
```

```python
MODEL = "PASTE_MODEL_HERE"

response = client.chat.completions.create(
    model=MODEL,
    messages=[{"role": "user", "content": "Prove that there are infinitely many primes."}],
)

print(response.choices[0].message.content)
```
